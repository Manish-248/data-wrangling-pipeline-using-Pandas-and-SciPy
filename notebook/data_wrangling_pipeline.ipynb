## 🔹 Handling Large Datasets

In real-world scenarios, datasets may be too large to fit into memory.

To handle such datasets efficiently, we use:
- Chunking (loading data in parts)
- Iterators
- Streaming techniques

This allows us to process data without loading everything at once.

```python
# ============================================================
# 📌 Load Dataset in Chunks using chunksize
# ============================================================

column_names = [
    'sepal_length',
    'sepal_width',
    'petal_length',
    'petal_width',
    'species'
]

iris_chunks = pd.read_csv(
    'iris.csv',
    header=None,
    names=column_names,
    chunksize=10   # Load 10 rows at a time
)

# Iterate through chunks
for i, chunk in enumerate(iris_chunks, start=1):
    print(f"\n🔹 Chunk {i} - Shape: {chunk.shape}")
    print(chunk.head())

    # For demo, break after 2 chunks
    if i == 2:
        break
```

```python
# ============================================================
# 📌 Iterator-Based Loading
# ============================================================

iris_iterator = pd.read_csv(
    'iris.csv',
    header=None,
    names=['C1', 'C2', 'C3', 'C4', 'C5'],
    iterator=True
)

# Fetch chunks dynamically
print("\n🔹 First 10 rows:\n", iris_iterator.get_chunk(10))
print("\n🔹 Next 20 rows:\n", iris_iterator.get_chunk(20))
print("\n🔹 Next 2 rows:\n", iris_iterator.get_chunk(2))
```

```python
# ============================================================
# 📌 Streaming Data using csv module
# ============================================================

import csv

with open('iris.csv', 'rt') as data_stream:
    for n, row in enumerate(csv.reader(data_stream)):
        if n == 0:
            print("\n🔹 First Row:", row)
        else:
            break
```

```python
# ============================================================
# 📌 Custom Batch Reader Generator
# ============================================================

def batch_read(filename, batch=5):
    """
    Reads a file in batches (chunks) and yields data as NumPy arrays.
    Useful for large datasets and streaming.
    """

    import csv
    import numpy as np

    with open(filename, 'rt') as data_stream:
        batch_output = []

        for n, row in enumerate(csv.reader(data_stream)):

            if n > 0 and n % batch == 0:
                yield np.array(batch_output)
                batch_output = []

            batch_output.append(row)

        # Yield remaining data
        yield np.array(batch_output)


# Test the generator
for batch in batch_read('iris.csv', batch=3):
    print("\n🔹 Batch Output:\n", batch)
    break
```

## 🔹 Working with Databases using SQLite

In real-world applications, data is often stored in databases.

SQLite is:
- Lightweight
- Serverless
- Fast
- Fully supported in Python

In this section, we will:
1. Create a database
2. Insert data
3. Query data using SQL
4. Load results into Pandas

```python
# ============================================================
# 📌 Create SQLite Database and Table
# ============================================================

import sqlite3

# SQL queries
drop_query = "DROP TABLE IF EXISTS temp_data;"

create_query = """
CREATE TABLE temp_data (
    date INTEGER,
    city TEXT,
    temperature REAL,
    destination INTEGER
);
"""

# Connect to database (file-based)
connection = sqlite3.connect("example.db")

# Execute queries
connection.execute(drop_query)
connection.execute(create_query)

connection.commit()

print("Database and table created successfully!")
```

```python
# ============================================================
# 📌 Insert Data into Database
# ============================================================

data = [
    (20140910, "Rome", 80.0, 0),
    (20140910, "Berlin", 50.0, 0),
    (20140910, "Wien", 32.0, 1),
    (20140911, "Paris", 65.0, 0)
]

insert_query = "INSERT INTO temp_data VALUES (?, ?, ?, ?)"

connection.executemany(insert_query, data)
connection.commit()

print("Data inserted successfully!")
```

```python
# ============================================================
# 📌 Retrieve Data using SQL Query
# ============================================================

import pandas as pd

query = """
SELECT date, city, temperature, destination
FROM temp_data
WHERE date = 20140910
"""

retrieved = pd.read_sql_query(query, connection)

print("\n🔹 Retrieved Data:\n")
print(retrieved)
```

```python
# ============================================================
# 📌 Close Database Connection
# ============================================================

connection.close()

print("Database connection closed.")
```

## 🔹 Combining DataFrames

In real-world scenarios, data often comes from multiple sources.

We use:
- Concatenation (pd.concat)
- Merging (pd.merge)

To:
- Combine rows or columns
- Join datasets based on keys

```python
# ============================================================
# 📌 Creating Sample DataFrames
# ============================================================

columns = ['name', 'age', 'gender', 'job']

user1 = pd.DataFrame([
    ['alice', 19, "F", "student"],
    ['john', 26, "M", "student"]
], columns=columns)

user2 = pd.DataFrame([
    ['eric', 22, "M", "student"],
    ['paul', 58, "F", "manager"]
], columns=columns)

user3 = pd.DataFrame({
    'name': ['peter', 'julie'],
    'age': [33, 44],
    'gender': ['M', 'F'],
    'job': ['engineer', 'scientist']
})

print("User1:\n", user1)
print("\nUser2:\n", user2)
print("\nUser3:\n", user3)
```

```python
# ============================================================
# 📌 Concatenate Rows (axis=0)
# ============================================================

users = pd.concat([user1, user2, user3], axis=0)

print("\n🔹 Combined Users (Row-wise):\n", users)
```

```python
# ============================================================
# 📌 Concatenate Columns (axis=1)
# ============================================================

height = pd.DataFrame({'height': [1.65, 1.80]})

combined_cols = pd.concat([user1, height], axis=1)

print("\n🔹 Combined Columns:\n", combined_cols)
```

```python
# ============================================================
# 📌 Merge DataFrames (Inner Join)
# ============================================================

user4 = pd.DataFrame({
    'name': ['alice', 'john', 'eric', 'julie'],
    'height': [165, 180, 175, 171]
})

merge_inner = pd.merge(users, user4, on='name')

print("\n🔹 Inner Join Result:\n", merge_inner)
```

```python
# ============================================================
# 📌 Merge DataFrames (Outer Join)
# ============================================================

merge_outer = pd.merge(users, user4, on='name', how='outer')

print("\n🔹 Outer Join Result:\n", merge_outer)
```

```python
# ============================================================
# 📌 Melt DataFrame (Wide → Long)
# ============================================================

melted = pd.melt(
    merge_outer,
    id_vars="name",
    var_name="variable",
    value_name="value"
)

print("\n🔹 Melted Data:\n", melted.head())
```

```python
# ============================================================
# 📌 Pivot DataFrame (Long → Wide)
# ============================================================

pivoted = melted.pivot(
    index='name',
    columns='variable',
    values='value'
)

print("\n🔹 Pivoted Data:\n", pivoted)
```

## 🔹 Handling Categorical Data

Categorical data represents labels or groups (e.g., weather, job roles).

Since machine learning models require numerical input, we convert categorical data into numerical form using:

- One-Hot Encoding (Dummy Variables)

This process creates binary columns for each category.

```python
# ============================================================
# 📌 Create Categorical Data
# ============================================================

categorical_feature = pd.Series([
    'sunny', 'cloudy', 'snowy', 'rainy', 'foggy'
])

print("\n🔹 Original Categorical Data:\n")
print(categorical_feature)
```

```python
# ============================================================
# 📌 One-Hot Encoding (pd.get_dummies)
# ============================================================

encoded = pd.get_dummies(categorical_feature)

print("\n🔹 One-Hot Encoded Data:\n")
print(encoded)
```

```python
# ============================================================
# 📌 Access Specific Encoded Category
# ============================================================

print("\n🔹 Encoding for 'sunny':\n")
print(encoded['sunny'])

print("\n🔹 Encoding for 'cloudy':\n")
print(encoded['cloudy'])
```

```python
# ============================================================
# 📌 One-Hot Encoding using Scikit-learn
# ============================================================

from sklearn.preprocessing import LabelEncoder, OneHotEncoder

# Step 1: Convert text → numeric labels
le = LabelEncoder()
labels = le.fit_transform(categorical_feature)

print("\n🔹 Label Encoded:\n", labels)

# Step 2: One-hot encoding
ohe = OneHotEncoder(sparse_output=False)

labels_reshaped = labels.reshape(-1, 1)
encoded_sklearn = ohe.fit_transform(labels_reshaped)

print("\n🔹 One-Hot Encoded (sklearn):\n")
print(encoded_sklearn)
```

```python
# ============================================================
# 📌 Apply Encoding to Iris Dataset (Target Column)
# ============================================================

encoded_target = pd.get_dummies(iris['target'])

print("\n🔹 Encoded Target Column:\n")
print(encoded_target.head())
```

## 🔹 Text Processing (NLP Basics)

Text data must be converted into numerical format before it can be used in machine learning.

We will use:
- CountVectorizer → word counts
- TF (Term Frequency) → normalized counts
- TF-IDF → importance of words
- N-grams → capture context

These techniques are widely used in NLP applications like chatbots, search engines, and sentiment analysis.

```python
# ============================================================
# 📌 Load Text Dataset (20 Newsgroups - Science Topics)
# ============================================================

from sklearn.datasets import fetch_20newsgroups

categories = ['sci.med', 'sci.space']

dataset = fetch_20newsgroups(categories=categories)

print("\n🔹 Sample Text:\n")
print(dataset.data[0][:500])  # Print first 500 characters
```

```python
# ============================================================
# 📌 Convert Text → Word Counts
# ============================================================

from sklearn.feature_extraction.text import CountVectorizer

count_vect = CountVectorizer()

word_count = count_vect.fit_transform(dataset.data)

print("\n🔹 Shape (documents, features):", word_count.shape)
```

```python
# ============================================================
# 📌 Inspect Words in First Document
# ============================================================

word_list = count_vect.get_feature_names_out()

for idx in word_count[0].indices:
    print(f"Word '{word_list[idx]}' appears {word_count[0, idx]} times")
```

```python
# ============================================================
# 📌 Term Frequency (Normalized)
# ============================================================

from sklearn.feature_extraction.text import TfidfVectorizer

tf_vect = TfidfVectorizer(use_idf=False, norm='l1')

word_freq = tf_vect.fit_transform(dataset.data)

word_list = tf_vect.get_feature_names_out()

print("\n🔹 Word Frequencies (First Document):\n")

for idx in word_freq[0].indices:
    print(f"Word '{word_list[idx]}' → Frequency {word_freq[0, idx]:.3f}")
```

```python
# ============================================================
# 📌 TF-IDF Vectorization
# ============================================================

tfidf_vect = TfidfVectorizer()

word_tfidf = tfidf_vect.fit_transform(dataset.data)

word_list = tfidf_vect.get_feature_names_out()

print("\n🔹 TF-IDF Values (First Document):\n")

for idx in word_tfidf[0].indices:
    print(f"Word '{word_list[idx]}' → TF-IDF {word_tfidf[0, idx]:.3f}")
```

```python
# ============================================================
# 📌 Unigrams (Single Words)
# ============================================================

documents = [
    'we love data science',
    'data science is hard'
]

from sklearn.feature_extraction.text import CountVectorizer

vect_uni = CountVectorizer(ngram_range=(1, 1))
uni_counts = vect_uni.fit_transform(documents)

print("\n🔹 Unigrams:\n", vect_uni.get_feature_names_out())
```

```python
# ============================================================
# 📌 Bigrams (Two-word combinations)
# ============================================================

vect_bi = CountVectorizer(ngram_range=(2, 2))
bi_counts = vect_bi.fit_transform(documents)

print("\n🔹 Bigrams:\n", vect_bi.get_feature_names_out())
```

```python
# ============================================================
# 📌 Uni + Bi-grams
# ============================================================

vect_mix = CountVectorizer(ngram_range=(1, 2))
mix_counts = vect_mix.fit_transform(documents)

print("\n🔹 Uni + Bi-grams:\n", vect_mix.get_feature_names_out())
```

```python
# ============================================================
# 📌 Hashing Vectorizer (Memory Efficient)
# ============================================================

from sklearn.feature_extraction.text import HashingVectorizer

hash_vect = HashingVectorizer(n_features=1000)

hashed = hash_vect.fit_transform(dataset.data)

print("\n🔹 Hashed Feature Shape:", hashed.shape)
```
